# **Operator**

```python
#Comparison Operator
a = 10
b = 20
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a == b)
print(a != b)
```

```python
#Logical operator
a = 10
b = 20
print(a > 5 and b > 15)
print(a > 5 or b < 15)
print(not(a > 5))

```

# **Conditionals**

```python
#Check the no is positive, negative or zero
n = int(input("n: "))
if n > 0:
  print("Positive")
elif n < 0:
  print("Negative")
else:
  print("0")
```

```python
#To check a no is odd or even
n = int(input())
if n % 2 == 0:
  print("Even")
else:
  print("Odd")
```

```python
#Check if leap or not
n = int(input())
if (n % 4 == 0 and n % 100 != 0) or n % 400 == 0:
  print("Leap")
else:
  print("Not leap")
```

```python
#Print even number from 1 to 100
for i in range(1, 101):
  if i % 2 == 0:
    print(i, end = " ")
```

```python
#Factorial
n = int(input())
fact = 1
for i in range(1, n + 1):
  fact = fact * i
print(fact)
```

```python
n = 1212
rev = 0
while n > 0:
  rem = n % 10
  rev = rev * 10 + rem
  n = n // 10
print(rev)
```

```python
n = 4153
count = 0
while n > 0:
  rem = n % 10
  count += 1
  n = n // 10
print(count)
```

#**Pattern Printing**

```python
n = int(input())
for i in range(1, n):
  print(" * " * i)
```

```python
n = int(input())
for i in range(1, n):
  for j in range(1, i + 1):
    print("*", end = " ")
  print()
```
